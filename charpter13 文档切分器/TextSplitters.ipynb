text_splitters内部的3个方法

split_text(self,text:str)->list[str]  该方法是抽象方法，具体的实现细节由子类来实现

create_documents(self,text:list[str],...)->list[Document] 底层调用了之前的split_text方法

split_documents(self,documents:Iterable[Document])->list[Document] 底层调用了之前的create_documents方法

CharacterTextSplitter的测试

```python
from  langchain_text_splitters import *
#示例文本
text=" langchain是一个用语言模型驱动的应用程序的框架，通过了1套工具和抽象，是开发者能够更容易得构建复杂的应用和程序"
#定义字符分割器
splitter=CharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=5,
    separator="",
)
#打印结果
texts=splitter.split_text(text)
for i ,chunk in enumerate(texts):
    print(f"{i+1}块 长度{len(chunk)}")
    print(chunk)
```

RecursiveCharacterSplitter 最常用

```python
from  langchain_text_splitters import RecursiveCharacterTextSplitter
#示例文本
text=" langchain\n\n是一个用语言模型驱动的应用程序的框架\n通过了1套工具和抽象\n是开发者能够更容易得构建复杂的应用和程序。文档分析场景示例：需要处理PDF/Word等格式"
#定义字符分割器
splitter=RecursiveCharacterTextSplitter(
    chunk_size=10,
    chunk_overlap=0,
    add_start_index=True,
    # separator="。",
)
#打印结果
texts=splitter.split_text(text)
for i ,chunk in enumerate(texts):
    print(f"{i+1}块 长度{len(chunk)}")
    print(chunk)
```
