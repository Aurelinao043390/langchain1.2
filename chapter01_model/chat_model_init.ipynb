```python
import time
from langchain.chat_models import init_chat_model
##统一使用init_chat_model（）完成模型的初始化
import os
from dotenv import load_dotenv
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from openai import max_retries

#读取，env文件中的配置信息，相关的环境变量以.env文件中的优先
load_dotenv(override=True)
my_api_key=os.getenv("API_KEY")
my_base_url=os.getenv("BASE_URL")
model= init_chat_model(
    model="qwen3.8-max",
    model_provider="openai",
    api_key=my_api_key,
    base_url=my_base_url,
    timeout=30,
    max_retries=3,
)
```

```python
response=model.invoke("翻译汉字“你好")
content=response.content
additional_kwargs=response.additional_kwargs
token_usage=response.response_metadata["token_usage"]
print("content:",content,"\n")
print("additional_kwargs:",additional_kwargs,"\n")
print("token_usage",token_usage,"\n")
```

测试多轮对话

```python
conversations=[
    {"role":"system","content":"你是1个专业的AI助手"},
    {"role":"user","content":"你好我叫Aureliano"}
]
#第一次对话
response1=model.invoke(conversations).content
print("#第一次对话:",response1)
#开启第二次对啊胡
conversations.append({"role":"assistant","content":response1})
conversations.append({"role":"user","content":"你还记得我名字吗"})
response2=model.invoke(conversations).content
print(response2)

```

消息对象列表传递消息-第1次对话


```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

messages=[
    SystemMessage(content="你是一个专业的数学老师"),
    HumanMessage(content="请你简单一下介绍一下斐波那契数列"),

]
response3=model.invoke(messages).content
print(response3)
# 消息对象列表传递消息-第2次对话
messages.append(AIMessage(content=response3))
messages.append(HumanMessage(content="我刚才问了什么问题"))
response4=model.invoke(messages)
print(response4)
```
