```python
import time
from langchain.chat_models import init_chat_model
##统一使用init_chat_model（）完成模型的初始化
import os
from dotenv import load_dotenv
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from pydantic_core.core_schema import model_field
from rich import print as rp


#读取，env文件中的配置信息，相关的环境变量以.env文件中的优先
load_dotenv(override=True)
my_api_key=os.getenv("API_KEY")
my_base_url=os.getenv("BASE_URL")
model= init_chat_model(
    model="qwen3.8-max",
    model_provider="openai",
    api_key=my_api_key,
    base_url=my_base_url,
    timeout=90,
    max_retries=3,
)
for chunk in model.stream("帮我总结一下巨蟹座女孩子的性格"):
    print(chunk.content,end="",flush=True)
```

批量调用 batch
一次性接收所有请求

```python
messages=[
    "帮我总结一下巨蟹座女孩子的性格",
    "我是天蝎座的男生，和巨蟹座女孩相处应该注意哪些点，请列出至少5个注意点",
    "如果谈恋爱的话，天蝎座男生和巨蟹座的女生的星座匹配程度如何"
]
response=model.batch(messages)
for message in response:
    answer=message.content.replace("\n","").replace("*","")
    print(answer,end="\n",flush=True)
```
