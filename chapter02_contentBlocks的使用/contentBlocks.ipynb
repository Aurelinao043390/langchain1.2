```python
import base64

from openai.types.responses import response_shell_call_command_added_event


#创建1个图片转换函数
#将本地图片转换成Base64编码的data uri字符串，方便在文本中嵌入图片数据
def encode_image(img_path:str,img_type:str="jpeg"):
  with open(img_path, "rb") as f:

      return f"data:image/{img_type};base64,{base64.b64encode(f.read()).decode("utf-8")}"


```

```python



from langchain.chat_models import init_chat_model
##统一使用init_chat_model（）完成模型的初始化
import os
from dotenv import load_dotenv

from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

# 设置图像路径
img_path_old= "1.jpeg"
img_path_new= "2.png"
base64_image_old=encode_image(img_path_old)
base64_image_new=encode_image(img_path_new,"png")
#读取，env文件中的配置信息，相关的环境变量以.env文件中的优先
load_dotenv(override=True)
my_api_key=os.getenv("API_KEY")
my_base_url=os.getenv("BASE_URL")
model= init_chat_model(
    model="qwen3.8-max",
    model_provider="openai",
    api_key=my_api_key,
    base_url=my_base_url,
    timeout=30,
    max_retries=3,
)

# #第1种旧版本的调用
human_message_old=HumanMessage(
        content= [
            {"type":"text","text":"图片里的内容有什么"},
            # {"type":"image_url","image_url":{"url":base64_image} }
            {"type":"image_url","image_url":f"data:image/jpeg;base64,{base64_image_old}"}
        ]

)

# #标准的统一写法
human_message_new=HumanMessage(
        content_blocks= [
            {"type":"text","text":"图片里的内容有什么"},

            {
                "type":"image",
                #新版本的写法需要清楚data:image/jpeg;base64这串字符
                "base64":base64_image_new.replace("data:image/png;base64,",""),
                #指定图片格式
                "mime_type":"png"
            }
        ]

)
response=model.invoke([human_message_new])
print(response.content_blocks)
print("======================================================================")
print(response.content)
```
