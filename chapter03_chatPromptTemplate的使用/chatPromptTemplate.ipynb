ChatPromptTemplate的调用
1 invoke() 返回的类型是prompt_values.ChatPromptValue


```python
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate, \
    MessagesPlaceholder

#初始化
my_template=ChatPromptTemplate.from_messages(
    [
        ("system","你是1个友好的AI助手，你叫{name}"),
        ("human","我叫Aureliano,很高兴认识你"),
        ("assistant","我很好，谢谢，也很高兴认识你"),
        ("human","{user_input}")
    ]
)
#调用
result=my_template.invoke({"name":"66","user_input":"重庆今天的天气怎么样"})
print(type(result),"\n")
print(result)
```

结合大模型的调用

```python

from langchain.chat_models import init_chat_model
##统一使用init_chat_model（）完成模型的初始化
import os
from dotenv import load_dotenv
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate
#读取，env文件中的配置信息，相关的环境变量以.env文件中的优先
load_dotenv(override=True)
my_api_key=os.getenv("API_KEY")
my_base_url=os.getenv("BASE_URL")
#初始化大模型
model= init_chat_model(
    model="qwen3.8-max",
    model_provider="openai",
    api_key=my_api_key,
    base_url=my_base_url,
    timeout=30,
    max_retries=3,
)


#初始化提示词模版
my_template=ChatPromptTemplate.from_messages(
    [
        ("system","你是1个友好的AI助手，你叫{name}"),
        ("human","我叫Aureliano,很高兴认识你"),
        ("assistant","我很好，谢谢，也很高兴认识你"),
        ("human","{user_input}")
    ]
)
#调用提示词模板
# prompt_value=my_template.format(name="66",user_input="1+1=？")
prompt_value=my_template.format_messages(name="66",user_input="1+1=？")
# prompt_value=my_template.invoke({"name":"66","user_input":"1+1=？"})

#调用大模型
resonse=model.invoke(prompt_value)
print(resonse.content_blocks,"\n")

print(resonse.content)
```

partial() 参数预填充

```python
#原始模板
inner_prompt_template1=ChatPromptTemplate.from_messages([("system","u are a {role},target user is {audience}")])
inner_prompt_template2=ChatPromptTemplate.from_messages([("human","Hello, my name is {name}, how are you,{task}")])

my_prompt8=ChatPromptTemplate.from_messages(
    [
        inner_prompt_template1,
        inner_prompt_template2
    ]
)
prompt_value8_1=my_prompt8.invoke({"name":"Aureliano","task":"can u introduce the Provence ChongQing ?","audience":"visitor","role":"tour guid"})
prompt_value8_2=my_prompt8.invoke({"name":"Aureliano","task":"can u introduce the Provence BeiJing ?","audience":"visitor","role":"tour guid"})




print(prompt_value8_1)
print(prompt_value8_2)
```

```python

inner_prompt_template1=ChatPromptTemplate.from_messages([("system","u are a {role},target user is {audience}")])
inner_prompt_template2=ChatPromptTemplate.from_messages([("human","Hello, my name is {name}, how are you,{task}")])

template9=ChatPromptTemplate.from_messages(
    [
        inner_prompt_template1,
        inner_prompt_template2
    ]
)
#部分变量预填充 ,必须要设定1个新变量接收返回值，使用新变量来调用invoke函数
partial_template=template9.partial(name="Aureliano",audience="visitor",role="tour guid")

prompt_value9_1=partial_template.invoke({"task":"can u introduce the Provence ChongQing ?"})
prompt_value9_2=partial_template.invoke({"task":"can u introduce the Provence BeiJing ?"})


print(prompt_value9_1)
print(prompt_value9_2)
```

消息占位符的使用
多轮对话系统存储历史消息以及agent的中间步骤处理此功能很有用
方式1 消息对象列表形式使用placeholder

```python

template10=ChatPromptTemplate.from_messages(
    [
        ("system","你是1个AI助手"),
        ("placeholder","{conversation}")
    ]
)
#部分变量预填充 ,必须要设定1个新变量接收返回值，使用新变量来调用invoke函数
result10=template10.invoke({
        "conversation":[
           HumanMessage(content="你好,请问明天的天气如何"),
            AIMessage(content="重庆明天天气阴转小雨，气温12-19摄氏度，出门记得带伞"),
           HumanMessage(content="衣服穿几件比较合适")
        ]
    }
)
print(result10)
```

MessagesPlaceholder占位

```python
from langchain_core.prompts import MessagesPlaceholder

template11=ChatPromptTemplate.from_messages(
    [
        ("system","你是1个AI助手"),
        MessagesPlaceholder(variable_name="conversation")

    ]
)
#部分变量预填充 ,必须要设定1个新变量接收返回值，使用新变量来调用invoke函数
result11=template11.invoke({
        "conversation":[
           HumanMessage(content="你好,请问明天的天气如何"),
            AIMessage(content="重庆明天天气阴转小雨，气温12-19摄氏度，出门记得带伞"),
           HumanMessage(content="衣服穿几件比较合适")
        ]
    }
)
print(result11)
```
