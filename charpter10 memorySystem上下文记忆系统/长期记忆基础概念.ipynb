长期记忆的存储架构 分为4层 Store->Namespace->key->value

Store 记忆仓库 是Langgrah提供的，常用的实现类是inmemorysaver()和postgresStore

Namespace 命名空间 数据类型是任意长度的tuple[]表示的层级路径，用于给长期记忆分组和隔离，数据类型为字符串元组 存储的都是key-value的键值对，key可以唯一标识1条value

key 是namespace下的唯一标识，单条记忆的唯一键，数据类型为字符串str

value 存储的值，数据类型为字段dict[str,any]
