agent.invoke()是最基本的同步调用方法，但是它会阻塞程序执行直到返回最终结果

入参是字典类型{"message":[{"role":"system","content":"you are a helpful ai assistant"}]}

输出是完整的消息列表，被封装在字典中，是messages字段的值

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from langchain.agents import create_agent
import os
from rich import print as rprint
from langchain_core.messages import SystemMessage, HumanMessage

from tool import response

load_dotenv(override=True)
my_api_key=os.getenv("API_KEY")
my_base_url=os.getenv("BASE_URL")
#1 初始化大模型
myModel= init_chat_model(
    model="qwen3.8-max",
    model_provider="openai",
    api_key=my_api_key,
    base_url=my_base_url,
)
myAgent=create_agent(
    model=myModel
)
#2 调用agent
response=myAgent.invoke(
    {
        #严格校验，使用message少了1个s就会报错
        #
        # "messages":
        #     [
        #         SystemMessage("你是1个私人ai助手"),
        #         HumanMessage("请你介绍一下你自己")
        # ]


        # 没有定义角色的时候，会默认封装到human角色中

        "messages":
            [
            "你好"

        ]
    }
)
# rprint(response)
print(response)
```
