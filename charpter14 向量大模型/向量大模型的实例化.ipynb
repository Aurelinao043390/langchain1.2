模型的初始化

```python
import load_dotenv
import psycopg
import dashscope
import os
from dotenv import load_dotenv
from langchain.embeddings import init_embeddings
from langchain_openai import  OpenAIEmbeddings
load_dotenv(override=True)
my_api_key = os.getenv("ALL_API_KEY")
my_base_url = os.getenv("EMBEDDING_BASE_URL")
text = "通用多模态表征模型示例"
# 通用初始化
my_embedding_model=init_embeddings(
    model="qwen3.7-text-embedding",
    provider="openai",
    api_key=my_api_key,
    base_url=my_base_url,
)
response=my_embedding_model.embed_documents(["通用多模态表征模型示例"])

# print(resp)
```

句子的向量化

```python
import load_dotenv
import psycopg
import dashscope
import os
from dotenv import load_dotenv
from langchain.embeddings import init_embeddings
from langchain_openai import  OpenAIEmbeddings
load_dotenv(override=True)
my_api_key = os.getenv("ALL_API_KEY")
my_base_url = os.getenv("BASE_URL")

text = "通用多模态表征模型示例"
input = [{'text': text}]
# 百炼云的初始化
resp = dashscope.MultiModalEmbedding.call(
    model="qwen3-vl-embedding",
    api_key=my_api_key,
    BASE_URL=my_base_url,
    input=text
)
print(resp)

```
