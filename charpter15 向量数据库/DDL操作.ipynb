```python
from pymilvus import MilvusClient
# client = MilvusClient(host='localhost', port=19530)
client = MilvusClient("http://localhost:19530")
```

查看数据库

```python
existed_database=client.list_databases()
for db in existed_database:
    print(db)
```

创建数据库

```python
db_name="rag_db"
if db_name not in existed_database:
    client.create_database(db_name)
print(existed_database)
```

删除数据库

```python
client.drop_database(db_name)
```

切换数据库

```python
vectorDB="vector_db"
if vectorDB not in existed_database:
    client.create_database(vectorDB)
```

```python
client.use_database(db_name=vectorDB)
```

查看数据库下的Collections

```python
client.use_database(db_name=vectorDB)
collections=client.list_collections()
print(collections)
```

创建Collections

```python
collection_name="docs"
client.create_collection(
    collection_name=collection_name,
    dimension=1024,#必须和使用的嵌入向量模型的维度相同
    metric_type="COSINE",#向量夹角计算方式，余弦计算


)
```

删除Collections

```python
client.drop_collection(collection_name=collection_name)
```
